```python


# Data Preprocessing
# Training Image preprocessing
print("🍎 Loading Training Dataset...")

try:
    # ✅ Keep your original code exactly as-is
    training_set = tf.keras.utils.image_dataset_from_directory(
        '/Users/kotireddy/Desktop/deepfritveg_identificatiion/fruit veg data set master/train',
        labels="inferred",
        label_mode="categorical",
        class_names=None,
        color_mode="rgb",
        batch_size=32,
        image_size=(64, 64),
        shuffle=True,
        seed=None,
        validation_split=None,
        subset=None,
        interpolation="bilinear",
        follow_links=False,
        crop_to_aspect_ratio=False
    )

    # ✅ Print verification details
    print("\n✅ Training Dataset Loaded Successfully!")
    print("📁 Classes Found:", len(training_set.class_names))
    print("📋 Class Names:", training_set.class_names)
    print("🖼️ Image Size:", training_set.element_spec[0].shape[1:3])
    print("📦 Batch Size:", training_set.element_spec[0].shape[0])
    print("🔢 Total Batches:", tf.data.experimental.cardinality(training_set).numpy())

except Exception as e:
    print("\n❌ Error Loading Dataset:")
    print(e)
```

```python
# Validation Image Preprocessing
print("\n📂 Loading Validation Dataset...")

try:
    validation_set = tf.keras.utils.image_dataset_from_directory(
        '/Users/kotireddy/Desktop/deepfritveg_identificatiion/fruit veg data set master/validation',
        labels="inferred",
        label_mode="categorical",
        class_names=None,
        color_mode="rgb",
        batch_size=32,
        image_size=(64, 64),
        shuffle=True,
        seed=None,
        validation_split=None,
        subset=None,
        interpolation="bilinear",
        follow_links=False,
        crop_to_aspect_ratio=False
    )
     # ✅ Print verification details
    print("\n✅ validation Dataset Loaded Successfully!")
    print("📁 Classes Found:", len(training_set.class_names))
    print("📋 Class Names:", training_set.class_names)
    print("🖼️ Image Size:", training_set.element_spec[0].shape[1:3])
    print("📦 Batch Size:", training_set.element_spec[0].shape[0])
    print("🔢 Total Batches:", tf.data.experimental.cardinality(training_set).numpy())

except Exception as e:
    print("\n❌ Error Loading Dataset:")
    print(e)
```

```python
# Building Model
cnn = tf.keras.models.Sequential()
```

```python
# Building Convolution Layer

cnn.add(tf.keras.layers.Conv2D(filters=32,kernel_size=3,padding='same',activation='relu',input_shape=[64,64,3]))
cnn.add(tf.keras.layers.Conv2D(filters=32,kernel_size=3,activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2,strides=2))

try:
    print("\n🧠 Building Convolution Layers...")
    print("Adding Conv2D Layer 1 (filters=32, kernel=3x3, padding='same', activation='relu')...")
    print("Adding Conv2D Layer 2 (filters=32, kernel=3x3, activation='relu')...")
    print("Adding MaxPooling2D Layer (pool_size=2, strides=2)...")
    print("✅ Convolution layers built successfully!")

except Exception as e:
    print("❌ Error while building convolution layers:", e)
```

```python
cnn.add(tf.keras.layers.Dropout(0.5)) #To avoid overfitting
```

```python
cnn.add(tf.keras.layers.Flatten())

try:
    print("\n📄 Adding Flatten Layer...")
    print("Flattening the feature maps into a 1D vector for the dense layers...")
    print("✅ Flatten layer added successfully!")

except Exception as e:
    print("❌ Error while adding Flatten layer:", e)
```

```python

cnn.add(tf.keras.layers.Dense(units=128,activation='relu'))

try:
    print("\n⚡ Adding Dense Layer...")
    print("Setting units=128 with ReLU activation for feature learning...")
    print("✅ Dense layer (units=128, activation='relu') added successfully!")

except Exception as e:
    print("❌ Error while adding Dense layer:", e)
```

```python
# Output Layer
cnn.add(tf.keras.layers.Dense(units=36,activation='softmax'))

try:
    print("\n🎯 Adding Output Layer...")
    print("Setting units=36 with Softmax activation for multi-class classification...")
    print("✅ Output layer (units=36, activation='softmax') added successfully!")

except Exception as e:
    print("❌ Error while adding Output layer:", e)
```

```python
# Compiling and Training Phase

cnn.compile(optimizer='rmsprop',loss='categorical_crossentropy',metrics=['accuracy'])

try:
    print("\n🧮 Compiling the CNN model...")
    print("Using optimizer='rmsprop', loss='categorical_crossentropy', and metrics=['accuracy']...")
    print("Please wait while the model is being compiled...")
    print("✅ Model compiled successfully!")

except Exception as e:
    print("❌ Error while compiling the CNN model:", e)
```

```python

cnn.summary()
```

```python
# Training

training_history = cnn.fit(x=training_set,validation_data=validation_set,epochs=70)
```

```python
# Saving model
cnn.save('fruit_model.h5')
```
